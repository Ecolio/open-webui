```python
import pymupdf  # Importing the library as pymupdf
import os

def convert_pdf_to_markdown(pdf_path, output_path, images_folder="images"):
    # Create a folder for images
    os.makedirs(images_folder, exist_ok=True)
    
    # Open the PDF file
    document = pymupdf.open(pdf_path)
    
    # Initialize a variable to hold the Markdown content
    markdown_content = ""

    # Iterate through the pages
    for page_num in range(document.page_count):
        page = document.load_page(page_num)
        text = page.get_text("text")
        
        # Extract images
        image_list = page.get_images(full=True)
        for img_index, img in enumerate(image_list):
            xref = img[0]
            base_image = document.extract_image(xref)
            image_bytes = base_image["image"]
            image_ext = base_image["ext"]
            image_filename = f"{images_folder}/page_{page_num + 1}_image_{img_index + 1}.{image_ext}"
            with open(image_filename, "wb") as image_file:
                image_file.write(image_bytes)
            text += f"\n![Image](./{image_filename})\n"

        # Convert page text to Markdown (simple approach)
        markdown_content += f"# Page {page_num + 1}\n\n"
        markdown_content += text
        markdown_content += "\n\n---\n\n"  # Page separator in Markdown

    # Close the document
    document.close()

    # Write the Markdown content to a file
    with open(output_path, "w") as md_file:
        md_file.write(markdown_content)

# Example usage
pdf_path = "/Users/lijou/Documents/Documents/Project/Notes_Docs/PreAward/exemples/Binder1.pdf"
output_path = "/Users/lijou/Documents/Documents/Project/Notes_Docs/PreAward/exemples/pymupdf_test.md"
convert_pdf_to_markdown(pdf_path, output_path)
```
